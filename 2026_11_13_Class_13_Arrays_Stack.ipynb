- append : Insert at the end ✅

- pop: remove from end by default ✅
- len(list) == 0: empty()

````python
class Stack:
    def __init__(self):
        self.data = []

    def push(self, val): # overflow if we keep a limit on size
        self.data.append(val)
        
    def pop(self):
        self.data.pop()

    def empty(self):
        return len(self.data) == 0 # undeflow
````

**Question**  
https://leetcode.com/problems/valid-parentheses/

````python
# TC, SC: O(n)
class Solution:
    
    def isValid(self, s: str) -> bool:

        stack = []

        for c in s:
            if self.is_open(c):
                stack.append(c)
            elif self.is_close(c):
                if len(stack) == 0:
                    return False
                
                if not self.is_match(stack.pop(), c):
                    return False

        return len(stack) == 0
        
    
    def is_open(self, sym):
        return sym in ["{", "[", "("]

    def is_close(self, sym):
        return sym in ["}", "]", ")"]

    def is_match(self, open, close):
        brackets = {"{": "}", "(": ")", "[": "]"}
        return brackets[open] == close
    

        
````

## Replace each element with the next greater element on it's right
```
[4,3,2,5,6,7,1]
 5,5,5,6,7,-1,-1

[4,5,2,3,6]
 5,6,3,6,-1
```

````python

# [4,3,2,5,6,7,1]
#  5,5,5,6,7,-1,-1

# [4,5,2,3,6]
#  5,6,3,6,-1


# TC: O(n^2)
# SC: O(1)
def greater_on_right(arr):
    for i in range(len(arr)):
        res = -1
        for j in range(i+1, len(arr)):
            if arr[i] < arr[j]:
                res = arr[j]
                break
        arr[i] = res

    return arr

a = [4,3,2,5,6,7,1]
print(a, greater_on_right(a))
a = [4,5,2,3,6]
print(a, greater_on_right(a))


## Optimal Solution
# TC: O(n)
# SC: O(n)

def greater_on_right(arr):
    stack = []

    for i in range(len(arr)-1,-1,-1):
        curr = arr[i]

        while len(stack) > 0 and curr > stack[-1]:
            stack.pop()

        if len(stack) == 0:
            arr[i] = -1
        else:
            arr[i] = stack[-1]

        stack.append(curr)
            
    return arr

a = [4,3,2,5,6,7,1]
print(a, greater_on_right(a))
a = [4,5,2,3,6]
print(a, greater_on_right(a))
````

**Question**  
https://leetcode.com/problems/next-greater-element-i/  

Two approach:  
- Value based (R->L)  
- Index based (L->R)  

````python
# Value Based R->L
# TC: SC: O(n)
class Solution:
    def nextGreaterElement(self, nums1: list[int], nums2: list[int]) -> list[int]:
                
        stack = []
        d = {}

        for i in range(len(nums2)-1,-1,-1):
            curr = nums2[i]

            while len(stack) > 0 and curr > stack[-1]:
                stack.pop()

            if len(stack) == 0:
                d[curr] = -1
            else:
                d[curr] = stack[-1]

            stack.append(curr)
        
        res = []
        for v in nums1:
            res.append(d[v])
        
        return res
````
