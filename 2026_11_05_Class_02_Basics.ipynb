### Pallindrome

```python
s = input("enter a string:")
# simple: creates a copy of string and takes extra space
print(s == s[::-1])

# using loop
i = 0
j = len(s)-1
res = True
while i < j:
    if s[i] != s[j]:
        res = False
        break
    i+=1
    j-=1
print(res)
# abcdba
# 012345
# i 0 1 2
# j 5 4 3
    
```

### Number Pallindrome

```python
# requires reversing a number
def reverse_num(num):
    res = 0
    while num != 0:
        res = res*10 + num%10
        num = num//10
    return res

print(reverse_num(123))

# num 123 12   1   0
# res 0    3  32 321
```

```python
def fact(n):
    res = 1
    for i in range(1, n+1):
        res *=i
    return res

print(fact(5))
```
