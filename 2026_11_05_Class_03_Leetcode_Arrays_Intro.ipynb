```python
# update vs insert: update means replace. insert means add 
# keeping all other elements as is
a = [10,20,30,40]
val = 100
idx = 1
a.append(0) # add a dummy value to increase size
for i in range(len(a)-1, idx, -1):
    a[i] = a[i-1]
a[idx] = val

# [10 20 20 30 40 50]
#   0  1  2  3  4 5
# i 5 4 3 2

print(a)
```

## Simple array problems

1. Find max element is an array 
1. Find max element and min element in a single pass
1. Reverse an array in place
1. Built-in min, max, reverse
1. Reverse vowels in an array https://leetcode.com/problems/reverse-vowels-of-a-string/

```python
def get_max(arr): 
    if len(arr) == 0:
        return None
    res_max = arr[0] # assume first element is biggest
    res_min = arr[0]
    for val in arr:
        if val > res_max:
            res_max = val
        if val < res_min:
            res_min = val

    return res_min, res_max

print(get_max([1,5,3,2,4,8,3]))
```
