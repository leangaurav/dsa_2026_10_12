### Solution-1
Find position of first zero.  
Loop over all data swapping the non zero elements with the zero position while incrementing zero position index.
```
[0,10, 0,30,12]
 0  1  2  3  4
i=0 1 2 3
j=0 1
[10,10, 0,30,12]

[10,20, 0, 0,30]
  0  1  2  3  4

[0, 0, 0]

[10,20,30]


[0,10, 0, 0,30,12]
 0  1  2  3  4  5
first_zero=0 1 1 1 2 3
i=         0 1 2 3 4 5
[10,30 12  0 30 12]
```

````python
class Solution:
    def moveZeroes(self, nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # [0,1,0,3,12]
        # -1 0 1
        # 0 1 2
        # [10,20,0,30]
        #.  0 1  2  3
        i = 0 # pointer to zero element
        while i < len(nums):
            if nums[i] == 0:
                break
            i += 1

        j = i+1
        while j < len(nums):
            if nums[j] != 0:
                nums[i], nums[j] = nums[j], nums[i]
````

### Solution-2        

````python
class Solution:
    def moveZeroes(self, nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # [0,1,0,3,12]
        # -1 0 1
        # 0 1 2
        zero_idx = -1
        for i in range(len(nums)):
            if nums[i] != 0:
                nums[zero_idx], nums[i] = nums[i], nums[zero_idx]
                zero_idx += 1
            else:
                if zero_idx == -1:
                    zero_idx = i
````

````python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
        i = 0
        for j in range(1,len(nums)):
            if nums[j] != nums[i]:
                i += 1
                nums[i]=nums[j]
                
        return i + 1
````

## https://leetcode.com/problems/two-sum/

````python
## Brute force
## TC: O(n^2)
## SC: O(1)
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        # target = 9
        # [2,11,15,7]
        # {2:0, 11:1, 15:2 }

        for i in range(len(nums)):
            for j in range(i+1, len(nums)):
                if nums[i] + nums[j] == target:
                    return i, j

        return -1,-1
````

````python
# Optimal Solution
## TC: O(n)
## SC: O(n)
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        # target = 9
        # [2,11,15,7]
        # {2:0, 11:1, 15:2 }
        d = {}
        for i,val in enumerate(nums):
            diff = target-val
            if diff in d:
                return i, d[diff]
            d[val] = i
        
        return -1, -1 
````
