## Counting frequency

Basic XOR Ops:

```python
# https://leetcode.com/problems/single-number/

class Solution:
    def singleNumber(self, nums: list[int]) -> int:
        
        res = 0
        for v in nums: # TC: O(n)
            res = v ^ res
        return res    

singleNumber([5,6,5])

# TC: O(n)
# SC: O(1)

#    res = 0000  
# [ 5     6      5]
#   0101  0110   0101
#   v 5     6
# res 0101  0011  0110
```

```python
# https://leetcode.com/problems/missing-number/

# Solution-1
# TC: O(n)
# SC: O(1)
class Solution:
    def missingNumber(self, nums: list[int]) -> int:

        total = 0
        for v in nums:
            total += v
        
        n = len(nums)
        expected = n * (n+1)//2
        return expected - total


# Solution-2
# TC: O(n)
# SC: O(1)
class Solution:
    def missingNumber(self, nums: list[int]) -> int:
        n = len(nums)
        return n * (n+1)//2 - sum(nums)


# Solution-3
# TC: O(n)
# SC: O(1)
class Solution:
    def missingNumber(self, nums: list[int]) -> int:
        res = 0
        for i in range(len(nums) + 1):
            res = res ^ i
        
        for v in nums:
            res = res ^ v
        return res
```
